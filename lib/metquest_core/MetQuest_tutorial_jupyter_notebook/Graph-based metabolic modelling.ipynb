## Task 1: Constructing individual and community graphs from GSMMs

**_Learning Outcomes_**:
Construction of bipartite graphs of individual and community of organisms from their respective genome-scale metabolic models.

**Overview**:

A metabolic network can be represented as a directed bipartite graph **_G(M, R, E)_**, where **_M_** is the set of metabolites in the metabolic network whereas **_R_** is the set of reactions and **_E_** the set of edges in the network. A directed edge always connects a metabolite node with the reaction node or reaction node to product metabolite nodes. Reversible reactions are denoted with two different reaction identifiers, for forward and reverse reactions.

These metabolic networks capture all the metabolism from uptake reactions to biomass reactions and secretion of product metabolites from the genome of the organism.

![Metabolic network of individual organism (Ravikrishnan, A., (2020))](individual.jpg)

Metabolic network of individual organism (Ravikrishnan, A., (2020))

This step can be executed in MetQuest using a class called ```construct_graph```. This class contains the required functions to create a metabolic network of either an individual orgainsm (when ```number_of_xml=1```) or a community containing more than one organism (when ```number_of_xml>1```) based on the requirement. Here, ```number_of_xml``` is one of the arguments passed into the function along with the ```path_to_models```.

Final output will be a NetworkX DiGraph Object and a ```name_map``` dictionary with keys as reaction ids in graph to the reaction ids in the model. 

**Note**: The graphs are stored as .gpickle files in the model path which can be loaded as graphs later using ```networkx.read_gpickle(filename)```.

## Task 2: Simulating the metabolic networks in a growth medium

**_Learning Outcomes_**: Simulating and studying the metabolic capability of a community or an individual organism.

_In silico_ simulation of the organism in a given nutrient medium (seed metabolites) can be done using the ```forward_pass``` function in MetQuest. It takes the corresponding NetworkX DiGraph Object and the nutrient media components as input. Individual organism graph can be supplied for simulating individual organisms, community metabolic graph can be given for simulating a community.

At the end of this step, we will have 
* ```lower_bound_metabolite``` - A dictionary containing minimum number of steps required to reach the metabolites in the network.
* ```status_dict``` - A dictionary containing all the reactions that are visited for the given nutrient medium (seed metabolites).
* ```scope``` - A set containing all the metabolites that can be produced by the network for the given nutrient medium (seed metabolites). 

These information about an individual organism or a community can be exploited in many different ways to understand the metabolism.

### Simulating individual organism

````python
# Loading the individual graph object from the gpickle file

import networkx as nx
import metquest

graph_pic = path_to_models+'/P1_.gpickle' # Graph of P. stipitis
G_pic = nx.read_gpickle(graph_pic)


# Loading the seed metabolites file for serving as the nutrient media

seed_met_file = path_to_models+'/seed_mets.txt'
with open(seed_met_file) as f:
    seed_metabolites = f.read().splitlines()
seed_metabolites = set([i for i in seed_metabolites if i.find('S1')!=0])
````

````python
lowboundmet_pic, status_dict_pic, scope_pic = metquest.forward_pass(G_pic, seed_metabolites)
````

**Note**: The outputs below will be based on reaction ids and metabolite ids. If we want to learn about the details such as which reaction id/metabolite id points to which reaction/metabolite, all these details will be available in the [BiGG Database](http://bigg.ucsd.edu/) and also in the models which can be extracted by writing a Python script.

````python
# Printing first 20 entries of lowboundmet_pic dictionary

{k:lowboundmet_pic[k] for k in list(lowboundmet_pic)[:20]}
````

````python
# Printing first 20 entries of status_dict_pic dictionary

{namemap_individual[k]:status_dict_pic[k] for k in list(status_dict_pic)[:20]}
````

````python
scope_pic
````

````python
# Number of reactions visited in P. stipitis individual network for the given media

visited_pic = {k:status_dict_pic[k] for k in list(status_dict_pic) if status_dict_pic[k]=='V'}
print('Number of reactions visited in P. stipitis individual network: %d' % len(visited_pic))
````

### Simulating a community

````python
# Loading the commuity graph object from the gpickle file

import networkx as nx

# Graph of P. stipitis - S. cerevisiae community
graph_community = path_to_models+'/S1_P1_.gpickle'
G_community = nx.read_gpickle(graph_community)


# Loading the seed metabolites file for serving as the nutrient media

seed_met_file = path_to_models+'/seed_mets.txt'
with open(seed_met_file) as f:
    seed_metabolites = f.read().splitlines()
seed_metabolites = set(seed_metabolites)
````

````python
lowboundmet_community, status_dict_community, scope_community = metquest.forward_pass(G_community, seed_metabolites)
````

````python
# Printing first 20 entries of status_dict_community dictionary

{namemap_comm[k]:status_dict_community[k] for k in list(status_dict_community)[:20]}
````

**Note**: The outputs below will be based on reaction ids and metabolite ids. If we want to learn about the details such as which reaction id/metabolite id points to which reaction/metabolite, all these details will be available in the [BiGG Database](http://bigg.ucsd.edu/) and also in the models which can be extracted by writing a Python script.

````python
# Number of reactions visited in P. stipitis - S. cerevisiae community individual network for the same media

visited_community = {k:status_dict_community[k] for k in list(status_dict_community) if status_dict_community[k]=='V'\
                     and k.find('Org_P1')>=0}
print('Number of reactions visited in P. stipitis in community network: %d' % len(visited_community))
````

### Inference

````python
print('Number of reactions visited in P. stipitis individual network: %d' % len(visited_pic))
print('Number of reactions visited in P. stipitis in community network: %d' % len(visited_community))
````

It can be seen that the number of reactions visited in _P. stipitis_ is more in community than when simulated individually. This increase in metabolic capability is due to the exchange of metabolites produced by _S. cerevisiae_ which were previously not available to _P. stipitis_. This small experiment proves the possibility of interactions existing between _P. stipitis_ and _S. cerevisiae_ when they are present together.

## Task 3: Finding pathways in the metabolic network

**_Learning Outcomes_**: 
1. Finding subnetworks/pathways in metabolic networks of microbial community and in an individual organism. 
2. Application of such methods to study the exchange of metabolites in a counity.

Pathways are essentially sub-networks in the Genome-scale metabolic network of an organism or a community. Identifying these pathways can help us better understand the role of these pathways in the synthesis of key desired metabolites. In case of community models, this can help in extracting information abou the excahnge of metabolites.

We are using the ```find_pathways``` function in MetQuest to perform this task.
Input:
1. NetworkX DiGraph Object on which the pathways are to be found
2. The nutrient media components that will be the start of every pathway found and 
3. The pathway length cutoff which is the maximum length of the pathway

Output:
1. ```pathway_table``` – Dictionary of dictionary containing the pathways of different sizes identified for every metabolite. This will have only the acyclic/ branched pathways.
2. ```cyclic_pathways``` – Dictionary of dictionary containing cyclic pathways of different sizes identified for every metabolite.
3. ```scope``` – Set of metabolites which can be synthesised.

````python
pathway_table, c_path, scope = metquest.find_pathways(G_community, seed_metabolites, 10)
````

**Note**: The outputs below will be based on reaction ids and metabolite ids. If we want to learn about the details such as which reaction id/metabolite id points to which reaction/metabolite, all these details will be available in the [BiGG Database](http://bigg.ucsd.edu/) and also in the models which can be extracted by writing a Python script.

````python
pathway_table
````

**I have extracted a pathway of length 9 from the `pathway_table`. Let us visualize the network.**

The extracted pathway is this:
`['Org_P1 ERR121','Org_P1 IR426','Org_P1 IR436','Org_P1 IR611','Org_P1 IR718','Org_P1 IR729','Org_P1 IR730','Org_S1 ER121','Org_S1 IR865']`.
These are the reaction nodes, their corresponding substrates need to be extracted to plot a required bi-partite graph of the extracted pathway.

````python
initial_nodes = ['Org_P1 ERR121','Org_P1 IR426','Org_P1 IR436','Org_P1 IR611','Org_P1 IR718',\
                'Org_P1 IR729','Org_P1 IR730','Org_S1 ER121','Org_S1 IR865']
subgraph_nodes = []
for n in initial_nodes:
    subgraph_nodes.append(n)
    for i in G_community.predecessors(n):
        subgraph_nodes.append(i)
for i in G_community.successors(n):
    subgraph_nodes.append(i)

extracted_subgraph = G_community.subgraph(subgraph_nodes)
````

Plotting the graph:

````python
import matplotlib.pyplot as plt
from pyvis.network import Network
net = Network(notebook=True, directed=True)
net.from_nx(extracted_subgraph)
net.show('example.html')
````

**Legend**:

Org_P1 = Organism _Pichia stipitis_

Org_S1 = Organism _Saccharomyces cerevisiae_

It can be seen from the graph that `S1 thm_e` is tranferred to `P1 thm_e`. This shows the metabolic exchange of thiamin (thm_e) from _Saccharomyces cerevisiae_ to _Pichia stipitis_.
